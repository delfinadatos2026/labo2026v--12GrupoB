```python

%pip install papermill

# %%
from pathlib import Path
import nbformat
import papermill as pm

# Carpeta donde está el workflow original
carpeta = Path("/content/buckets/b1/exp")

workflow = "Carlos_EXP_A_z621_WorkFlow_01_junior_grupoB_BO.ipynb"

notebook_entrada = carpeta / workflow

# Cinco semillas
semillas = [
    611147, 179213, 607681, 618271, 897563
]

# Carpeta donde se guardarán los notebooks ejecutados
salida = carpeta / "ejecutados_carlos"
salida.mkdir(parents=True, exist_ok=True)

# Primer número de experimento
experimento_base = 9550

# Leer el notebook original
nb_original = nbformat.read(notebook_entrada, as_version=4)

# Identificar la celda de parámetros
indice_parametros = None

for i, celda in enumerate(nb_original.cells):
    if celda.cell_type == "code" and "semilla_lanzador" in celda.source and "experimento_lanzador" in celda.source:
        indice_parametros = i
        break

if indice_parametros is None:
    raise ValueError("No se encontró la celda de parámetros.")

# Ejecutar el Workflow A para cada semilla
for i, semilla in enumerate(semillas):

    experimento = experimento_base + i

    # Crear una copia independiente del notebook original
    nb = nbformat.from_dict(nb_original)

    # Reemplazar los parámetros en la celda correspondiente
    celda = nb.cells[indice_parametros]

    celda.source = celda.source.replace(
        "semilla_lanzador <- 611147L",
        f"semilla_lanzador <- {semilla}L"
    )

    celda.source = celda.source.replace(
        "experimento_lanzador <- 9550L",
        f"experimento_lanzador <- {experimento}L"
    )

    # Guardar notebook temporal de entrada
    notebook_temporal = salida / f"temporal_EXP_A_{experimento}.ipynb"
    nbformat.write(nb, notebook_temporal)

    # Nombre del notebook ejecutado
    notebook_salida = salida / (
        f"{Path(workflow).stem}"
        f"_semilla_{semilla}"
        f"_exp_{experimento}.ipynb"
    )

    print(
        f"\nEjecutando Workflow A | "
        f"Semilla: {semilla} | "
        f"Experimento: {experimento}"
    )

    # Ejecutar sin pasar parámetros por Papermill
    pm.execute_notebook(
        str(notebook_temporal),
        str(notebook_salida),
        progress_bar=False,
        kernel_name="ir"
    )

    # Eliminar notebook temporal
    notebook_temporal.unlink()

print("\nFinalizaron los cinco experimentos del Workflow A.")
```
