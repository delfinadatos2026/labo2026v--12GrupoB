```python
# %%
!pip install papermill

# %%
from pathlib import Path
import papermill as pm

# Carpeta donde están los notebooks originales
carpeta = Path("/content/buckets/b1/exp")

# Los tres workflows originales
workflows = [
    "EXP_B_z621_WorkFlow_01_junior_grupoB_BO.ipynb",
    "EXP_A_z621_WorkFlow_01_junior_grupoB_BO.ipynb"
]

# Cinco semillas CAMBIAR POR LAS DE CADA UNO
semillas = [
    200017, 323503, 543121, 898117, 726769
]

# Carpeta independiente para los notebooks ejecutados
salida = carpeta / "ejecutados_delfina"
salida.mkdir(parents=True, exist_ok=True)

# Primer número de experimento
experimento_base = 9230

# Ejecutar los tres workflows para cada semilla
for i, semilla in enumerate(semillas):

    for j, workflow in enumerate(workflows):

        experimento = experimento_base + i * len(workflows) + j

        notebook_entrada = carpeta / workflow

        notebook_salida = salida / (
            f"{Path(workflow).stem}"
            f"_semilla_{semilla}"
            f"_exp_{experimento}.ipynb"
        )

        print(
            f"\nEjecutando {workflow} | "
            f"Semilla: {semilla} | "
            f"Experimento: {experimento}"
        )

        pm.execute_notebook(
            str(notebook_entrada),
            str(notebook_salida),
            parameters={
                "semilla_lanzador": semilla,
                "experimento_lanzador": experimento
            },
            progress_bar=False,
            kernel_name='ir'
        )

print("\nFinalizaron los 15 experimentos.")
```
